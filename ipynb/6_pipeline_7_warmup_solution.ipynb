Six used cars, each feature missing one value. Build a `Pipeline`
that imputes the median, standardizes, and ends with
`LinearRegression`. Fit it to predict `price` from `age` and `miles`,
then print the predictions and the RMSE.

Run the setup cell first.

````python
# Setup: imports and the data
# data frames and arrays
import pandas as pd
import numpy as np
# pipeline, imputer, scaler, model, error metric
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

# six cars, one missing value in age and in miles
cars = pd.DataFrame({
    "age":   [2, 5, np.nan, 8, 3, 6],
    "miles": [20000, 60000, 45000, np.nan, 30000, 75000],
    "price": [24000, 15000, 17000, 9000, 21000, 12000],
})
print(cars)
````

Expected output:

```
   age    miles  price
0  2.0  20000.0  24000
1  5.0  60000.0  15000
2  NaN  45000.0  17000
3  8.0      NaN   9000
4  3.0  30000.0  21000
5  6.0  75000.0  12000
```

## Solution

````python
# predict price from age and miles
X = cars[["age", "miles"]]
y = cars["price"]

# impute, scale, predict
pipe = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler()),
    ("predict", LinearRegression()),
])
# fit all three steps
pipe.fit(X, y)
# predict the same six cars
predictions = pipe.predict(X)
print(np.round(predictions, 0))
# root mean squared error
print(f"RMSE: {root_mean_squared_error(y, predictions):,.0f}")
````

Expected output:

```
[24097. 15195. 16003.  9257. 21309. 12138.]
RMSE: 451
```

The RMSE is the error on the six cars the model was trained on: on
new cars, it will be larger.
