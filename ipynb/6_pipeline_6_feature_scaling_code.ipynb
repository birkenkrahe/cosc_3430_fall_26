## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

housing_full = pd.read_csv("https://tinyurl.com/housing-csv")
housing_full["income_cat"] = pd.cut(
    housing_full["median_income"],
    bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],
    labels=[1, 2, 3, 4, 5])
strat_train_set, strat_test_set = train_test_split(
    housing_full, test_size=0.2,
    stratify=housing_full["income_cat"],
    random_state=42)
for df in (strat_train_set, strat_test_set):
    df.drop("income_cat", axis=1, inplace=True)
housing = strat_train_set.drop("median_house_value", axis=1)
housing_labels = strat_train_set["median_house_value"].copy()
housing_num = housing.select_dtypes(include=[np.number])
print(housing.shape, housing_num.shape)
```

## Code-along 1. Compare the spread of two features

```python
print(housing.describe()[["total_rooms","median_income"]]) # 2d array
```

## Code-along 2. Min-max scaling: `MinMaxScaler`

```python
from sklearn.preprocessing import MinMaxScaler

min_max_scaler = MinMaxScaler(feature_range=(-1,1))
housing_num_min_max_scaled = min_max_scaler.fit_transform(housing_num)

col = "total_rooms"
idx = housing_num.columns.get_loc(col)  # complicated because we're in pandas now
before = housing_num[col].describe()
after = pd.Series(housing_num_min_max_scaled[:,idx]).describe()
print(pd.DataFrame({"before":before, "after":after}))
```

## Code-along 3. Min-max scaling changes the scale, not the shape

```python
fig, axes = plt.subplots(1,2,figsize=(10,4))
housing_num[col].hist(ax=axes[0],bins=50)
pd.Series(housing_num_min_max_scaled[:,idx]).hist(ax=axes[1],bins=50)
plt.tight_layout()
plt.savefig("plot.png")
```
