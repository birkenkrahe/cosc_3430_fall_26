## Code-along 1. Fill in the confusion matrix

```python
TP = 40   # flagged AND actually sick
FN = 10   # sick but not flagged
FP = 40   # flagged but actually healthy
TN = 910  # healthy and correctly cleared

print(f"TP + FN = {TP + FN} (all sick: 50)")  # actual positives (P), the row sum
print(f"FP + TN = {FP + TN} (all healthy: 950)")  # actual negatives (N), the row sum
print(f"TP + FP = {TP + FP} (all flagged: 80)")  # predicted positives, the column sum
```

## Code-along 2. Accuracy

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)
print(f"Accuracy: {accuracy}")
```

## Code-along 3. The lazy model

```python
TP0, FN0, FP0, TN0 = 0, 50, 0, 950
print(f"Lazy accuracy: {(TP0 + TN0) / 1000}")
print(f"Sick patients caught: {TP0} of {TP0 + FN0}")
```

## Code-along 4. Precision

```python
precision = TP / (TP + FP)
print(f"Precision: {precision}")
```

## Code-along 5. Recall

```python
recall = TP / (TP + FN)
print(f"Recall: {recall}")
```

## Code-along 6. The precision/recall trade-off

```python
TP2 = 48
FN2 = 50 - TP2
FP2 = 200 - TP2
TN2 = 950 - FP2

precision2 = TP2 / (TP2 + FP2)
recall2 = TP2 / (TP2 + FN2)

print(f"t = 0.5: precision {precision:.2f}, "
      f"recall {recall:.2f}")
print(f"t = 0.2: precision {precision2:.2f}, "
      f"recall {recall2:.2f}")
```

## Code-along 7. `confusion_matrix()`

```python
import numpy as np
from sklearn.metrics import (confusion_matrix, precision_score,
                             recall_score, f1_score,
                             classification_report)

y_true = np.array([0] * 950 + [1] * 50)    # 950 healthy, then 50 sick
y_pred = np.array([0] * 910 + [1] * 40     # healthy: TN, FP
                  + [0] * 10 + [1] * 40)   # sick: FN, TP

print(confusion_matrix(y_true, y_pred))
print(f"Precision: {precision_score(y_true, y_pred):.2f}")
print(f"Recall:    {recall_score(y_true, y_pred):.2f}")
```

## Code-along 8. `classification_report()`

```python
print(classification_report(y_true, y_pred, target_names=["Healthy",
                            "Disease"]))
```

## Code-along 9. F1 and the harmonic mean

```python
def f1(p, r):
    return 2 * p * r / (p + r)

print(f"Our model: F1 = {f1(precision, recall):.2f}")
print(f"sklearn:   F1 = {f1_score(y_true, y_pred):.2f}")

P, R = 1.0, 1 / 50
print(f"Extreme model: arithmetic mean {(P + R) / 2:.2f}, F1 {f1(P, R):.2f}")
```

## Code-along 10. Which models get a high F1?

```python
cases = [("high P, high R", 0.90, 0.85),
         ("low P,  high R", 0.25, 0.90),
         ("high P, low R ", 0.90, 0.20),
         ("low P,  low R ", 0.25, 0.20)]
for label, p, r in cases:
    print(f"{label}: P {p:.2f}  R {r:.2f}  F1 {f1(p, r):.2f}")
```
