and `num_pipeline` (impute, then scale) from the code-alongs.

```python
# toy data and num_pipeline from code-alongs 1 and 2
import pandas as pd
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# two numeric columns, one NaN each
toy = pd.DataFrame({
    "age":    [25, 32, np.nan, 40, 28],
    "income": [50000, 64000, 58000, np.nan, 72000],
})

# two named steps, applied in this order
num_pipeline = Pipeline([
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler()),
])

# fit and transform all steps in one call
toy_prepared = num_pipeline.fit_transform(toy)
print(toy_prepared.round(2))
```

## Text: `print()`

Inside `print()`, a pipeline is shown as text.

```python
print(num_pipeline)
```

## A pipeline with a predictor as the last step

The same diagram for a pipeline that ends with a predictor
(`make_pipeline` names the steps for you).

```python
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression

pipeline = make_pipeline(SimpleImputer(strategy="median"),
                         StandardScaler(),
                         LinearRegression())
pipeline
```
